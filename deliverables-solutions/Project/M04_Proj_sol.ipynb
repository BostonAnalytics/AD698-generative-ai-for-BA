# Module 4 Checkpoint Solution Guide

Module 4 is a readiness audit. The goal is to decide whether Steps 1-3 are complete enough to begin grounded generation in Module 5.

Do not treat this as a week off. Weak retrieval in Step 3 will produce weak answers in Step 4.


## 1. Check Required Files


```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path("/content/drive/MyDrive/project_sec10k_rag")
DATA_DIR = BASE_DIR / "data"
OUTPUTS_DIR = DATA_DIR / "outputs"
VECTOR_STORE_DIR = DATA_DIR / "vector_store"

required_files = [
    "config/companies.csv",
    "data/outputs/download_manifest.csv",
    "data/outputs/filing_metadata.csv",
    "data/outputs/extracted_item_metadata.csv",
    "data/outputs/cleaned_item_metadata.csv",
    "data/chunked/sec10k_chunks.parquet",
    "data/vector_store/embedding_manifest.json",
    "data/outputs/retrieval_audit_step3.csv",
]

checkpoint = pd.DataFrame(
    {
        "required_artifact": required_files,
        "exists": [(BASE_DIR / path).exists() for path in required_files],
    }
)

checkpoint["status"] = checkpoint["exists"].map({True: "ready", False: "missing"})
checkpoint
```

## 2. Summarize Retrieval Evidence

```python
audit_path = OUTPUTS_DIR / "retrieval_audit_step3.csv"

if audit_path.exists():
    retrieval_audit = pd.read_csv(audit_path)
    display(retrieval_audit.head(10))

    if "manual_judgment" in retrieval_audit.columns:
        display(retrieval_audit["manual_judgment"].value_counts(dropna=False).to_frame("count"))
else:
    print("No retrieval audit found. Run Step 3 retrieval checks before starting generation.")
```
