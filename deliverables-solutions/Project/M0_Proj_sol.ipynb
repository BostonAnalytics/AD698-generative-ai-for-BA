# SEC 10-K Financial RAG Project Build Plan

This solution guide turns the project overview into a concrete build plan. The full project is a domain-scoped financial RAG system over SEC 10-K filings from 2020 through the latest available filing date at runtime.

The project has four technical stages:

1. Choose one NAICS industry, select public companies, download every available 10-K for those companies from 2020 onward, and extract 10-K sections.
2. Clean the selected sections and inspect corpus coverage by company, year, and item.
3. Chunk the cleaned text, create embeddings, save the vector index, and audit retrieval quality.
4. Generate grounded answers, evaluate them, and connect filing evidence to market behavior.

The most important habit is reproducibility. Every notebook should save clear files into the same project folder so later notebooks can load them without guessing.
