## 1. Build a Slide Outline


```python
from pathlib import Path
import pandas as pd

# In Google Colab, this is the expected project folder.
# If you run locally, replace this path with Path.cwd() / "project_sec10k_rag".
BASE_DIR = Path("/content/drive/MyDrive/project_sec10k_rag")

DATA_DIR = BASE_DIR / "data"
OUTPUTS_DIR = DATA_DIR / "outputs"
REPORTS_DIR = BASE_DIR / "reports"

for folder in [DATA_DIR, OUTPUTS_DIR, REPORTS_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

print("Project folder:", BASE_DIR)

(REPORTS_DIR / "tables").mkdir(parents=True, exist_ok=True)

slide_outline = pd.DataFrame(
    [
        (1, "Title and Product Framing", "Industry, target user, and business problem"),
        (2, "System Architecture", "Pipeline from SEC filings to grounded answer"),
        (3, "Data Scope", "NAICS code, companies, years, and SEC Items"),
        (4, "Cleaning and Corpus Quality", "EDA table and data quality issues"),
        (5, "Chunking and Embeddings", "Chunk size, overlap, model choice, and index type"),
        (6, "Retrieval Evidence", "One top-k retrieval example with scores"),
        (7, "Generation Design", "Prompt, citation requirements, and refusal behavior"),
        (8, "Evaluation Results", "Relevance, faithfulness, citation accuracy, hallucination risk"),
        (9, "Market Analysis", "Yahoo Finance feature and filing-to-market connection"),
        (10, "Failure Cases", "What failed, why it matters, and next fix"),
        (11, "Recommendations", "Next sprint priorities for a real FinTech product"),
    ],
    columns=["slide_number", "slide_title", "evidence_to_show"],
)

slide_outline.to_csv(REPORTS_DIR / "tables" / "presentation_slide_outline.csv", index=False)
slide_outline
```
